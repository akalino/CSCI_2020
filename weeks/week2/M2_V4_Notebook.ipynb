```python
# String methods
```

```python
my_string_variable = "This will be my string."
print("My string variable.")
```

```python
upper_string_variable = my_string_variable.upper()
print(upper_string_variable)
```

```python
lower_string_value = my_string_variable.lower()
print(lower_string_value)
```

```python
# Checking if strings are all upper or lower
print(lower_string_value.islower())
```

```python
print(my_string_variable.isupper())
```

## 1) Classes vs Functions: Pros and Cons
### Functions are great when:
- the task is a clear input → output computation
- you don’t need to store changing state over time
- you want simple, testable building blocks

### Classes are helpful when:
- you need to combine **data + behavior** in one place
- you have a “thing” with state (fields) that changes over time
- you want a reusable model (objects) with methods

### Downsides of classes:
- more syntax / more moving parts
- can be overused (classes aren’t always needed)


### Quick intuition
- If you find yourself passing the same variables into many functions repeatedly, a class might help.
- If your code is a one-off calculation, a function is usually better.


## 2) Anatomy of a Class
A class models a **thing**:
- a class defines a blueprint
- an object is an instance of that blueprint
- objects have:
  - **attributes** (data)
  - **methods** (functions that act on the data)


### Typical structure
```python
class Name:
    def __init__(self, ...):
        self.x = x

    def do_something(self, ...):
        ...
```


### What is `self`?
- `self` refers to the **current object**
- `self.attribute` is data stored on the object
- When you call `obj.method()`, Python automatically passes `obj` as `self`


## 3) A First Useful Class Example: `Timer`
This example is intentionally small: it stores state and provides a few methods.

> Image placeholder (from slides): `PLACEHOLDER_m7_v1_simple_class.png`


````
import time

class Timer:
    """A tiny timer object that measures elapsed time."""
    def __init__(self, label: str = 'timer'):
        self.label = label
        self.start_time = None

    def start(self):
        self.start_time = time.time()
        return self

    def elapsed(self) -> float:
        if self.start_time is None:
            return 0.0
        return time.time() - self.start_time

t = Timer('demo').start()
time.sleep(0.2)
print('Elapsed seconds:', round(t.elapsed(), 3))
````

### Common mistakes to avoid
- Forgetting `self` in method parameters
- Writing `x = ...` instead of `self.x = ...` when you mean to store state
- Overcomplicating: using classes when simple functions would do

## 4) Inheritance: Base Classes and Child Classes
Inheritance models an **is-a** relationship:
- a child class inherits attributes/methods from a parent
- can add behavior
- can override behavior
- goal: reuse code, avoid duplication


### When a base class helps
Use a base class when multiple classes share:
- the same attributes
- the same method patterns

Keep base classes **small and focused**.


### `super()`
If a child class defines its own `__init__`, it should usually call the parent:
```python
super().__init__(...)
```
This ensures base attributes get initialized and avoids duplicated code.


## 5) End-to-End Example: `Account` → `SavingsAccount`
We keep this example minimal:
- base class stores shared fields + behavior
- child class adds one rule (override)

> Image placeholder (from slides): `PLACEHOLDER_m7_v3_full_example.png`


````
class Account:
    """A simple bank account with a balance."""
    def __init__(self, owner: str, balance: float = 0.0):
        self.owner = owner
        self.balance = float(balance)

    def deposit(self, amount: float) -> None:
        if amount <= 0:
            raise ValueError('Deposit must be positive')
        self.balance += float(amount)

    def withdraw(self, amount: float) -> bool:
        """Withdraw if possible; return True if success, else False."""
        if amount <= 0:
            return False
        if amount > self.balance:
            return False
        self.balance -= float(amount)
        return True

    def __repr__(self) -> str:
        return f"Account(owner={self.owner!r}, balance={self.balance:.2f})"

class SavingsAccount(Account):
    """A savings account that enforces a minimum balance."""
    def __init__(self, owner: str, balance: float = 0.0, min_balance: float = 0.0, **kwargs):
        super().__init__(owner=owner, balance=balance)
        self.min_balance = float(min_balance)

    def withdraw(self, amount: float) -> bool:
        """Override: enforce min_balance."""
        if amount <= 0:
            return False
        if self.balance - float(amount) < self.min_balance:
            return False
        self.balance -= float(amount)
        return True

a = Account('Ada', 100)
s = SavingsAccount('Linus', 100, min_balance=50)
print(a)
print(s)
print('Account withdraw 80:', a.withdraw(80), a)
print('Savings withdraw 80 (should fail):', s.withdraw(80), s)
print('Savings withdraw 40 (should work):', s.withdraw(40), s)
````

### Overriding methods (simple rule)
- Child can redefine a method with the same name.
- Override only when needed.
- Keep overrides small and readable.


### What is `**kwargs`?
- `**kwargs` collects extra named arguments into a dictionary.
- Useful for flexible optional settings or passing options through layers.
- Use sparingly in this course (readability first).

## 6) When NOT to Use Inheritance
- Avoid inheritance when the relationship is **has-a** (composition)
  - Example: a car *has an* engine
- Avoid deep inheritance chains
- If you can solve it with a function or a simple class, do that

## Why list comprehensions?
- Shorter than loop-based list building
- Often more readable when simple
- Common use cases:
  - transform values
  - filter values
  - transform + filter


## The template
- Basic: `new = [expr for x in iterable]`
- Filter: `new = [expr for x in iterable if condition]`
Read it as: “Build a list of expr for each x ...”


> Image placeholder: side-by-side loop vs comprehension.


````
# Loop version
nums = [1, 2, 3, 4]
squares_loop = []
for x in nums:
    squares_loop.append(x*x)

# Comprehension version
squares_comp = [x*x for x in nums]

print('loop:', squares_loop)
print('comp:', squares_comp)
````

## Transform pattern
Example: squares
`squares = [x*x for x in nums]`


## Filter pattern
`evens = [x for x in nums if x % 2 == 0]`
The if goes at the end.


````
nums = list(range(10))
evens = [x for x in nums if x % 2 == 0]
print(evens)
````

## Transform + filter
Example: clean up words and keep only long ones:
`clean = [w.strip().lower() for w in words if len(w.strip()) >= 5]`


````
words = ['  Apple ', 'kiwi', 'BANANA', ' pear  ', 'grape']
clean = [w.strip().lower() for w in words if len(w.strip()) >= 5]
print(clean)
````

## Comprehensions + range()
- `nums = [i for i in range(10)]`
- `squares = [i*i for i in range(10)]`


````
nums = [i for i in range(10)]
squares = [i*i for i in range(10)]
print(nums)
print(squares)
````

## Readability rules
- Avoid nested comprehensions early
- If it’s hard to read, use a normal loop
- Don’t use comprehensions for side effects


## Quick practice
- Convert a loop to a comprehension
- Predict the output
- Decide: comprehension or loop for readability?

## What is a dictionary?
- Stores labeled data: key → value
- Why dicts? lookup by meaning, not position; great for counts/records/mappings.


## Creating and accessing
```python
d = {'apples': 3, 'bananas': 1}
empty = {}
d['apples']   # KeyError if missing
d['apples'] = 4
```


````
d = {'apples': 3, 'bananas': 1}
print(d['apples'])
d['apples'] = 4
print(d)
````

## Avoiding KeyError
- Test membership: `if key in d:`
- Use get with default: `d.get(key, 0)`
- Rule of thumb: beginners should love get.


> Image placeholder: KeyError vs using in/get.


````
d = {'apples': 3}
key = 'oranges'
print('in check:', key in d)
print('get default:', d.get(key, 0))
````

## Iterating through a dictionary
- keys: `for k in d:` or `d.keys()`
- values: `d.values()`
- pairs: `d.items()`


````
d = {'apples': 3, 'bananas': 1}
for k, v in d.items():
    print(k, '->', v)
````

## Counting pattern (classic dict use)
Goal: build a frequency table.
Often implemented with get:
`d[key] = d.get(key, 0) + 1`


> Image placeholder: counting demo screenshot.


````
text = 'To be or not to be'
counts = {}
for w in text.lower().split():
    counts[w] = counts.get(w, 0) + 1
print(counts)
````

## Practical strings toolkit (no regex)
- Clean up: `strip`, `lower/upper`, `replace`
- Break apart / combine: `split`, `' '.join(words)`


````
s = '  Hello, World!  '
print('strip:', s.strip())
print('lower:', s.lower())
print('replace:', s.replace('World', 'Python'))
words = ['CSCI', '2020', 'is', 'fun']
print('join:', ' '.join(words))
````

## Strings as sequences (reminder)
- Indexing and slicing work on strings
- Strings are immutable: you create new strings when changing them


## Common mistakes

- Overwriting values accidentally
- Modifying dict while iterating
- Forgetting to normalize strings before counting

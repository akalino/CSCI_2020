## What is a sequence?
- An ordered collection of items.
- Examples: quiz scores, coordinates `(x, y)`, strings.
- Key idea: order matters and you can access items by index.


## Lists vs tuples (big picture)
- List `[ ]`: **mutable**, used for collections that grow/change.
- Tuple `( )`: **immutable**, used for fixed records (like coordinates).


## Creating lists and tuples
```python
nums = [10, 20, 30]
empty = []
pt = (4, 9)
empty_t = ()
```
One-item tuple gotcha:
```python
t = (5,)
```


````
nums = [10, 20, 30]
empty = []
pt = (4, 9)
empty_t = ()
t = (5,)
print(nums, empty, pt, empty_t, t)
````

## Indexing
- Indexing starts at 0: `nums[0]`
- Negative indexing counts from the end: `nums[-1]`


> Image placeholder: screenshot showing list indexing and negative indexing.


````
nums = [10, 20, 30]
print('nums[0] =', nums[0])
print('nums[-1] =', nums[-1])
````

## Slicing
- Slice syntax: `seq[start:stop]` (stop is not included)
- Examples: `nums[1:3]`, `nums[:2]`, `nums[2:]`
- Optional step preview: `seq[start:stop:step]`


> Image placeholder: diagram showing slice bounds and stop excluded.


````
nums = [10, 20, 30, 40, 50]
print('nums[1:3] =', nums[1:3])
print('nums[:2] =', nums[:2])
print('nums[2:] =', nums[2:])
print('nums[::2] =', nums[::2])
````

## Mutability
- Lists are mutable: `nums[0] = 999` (allowed)
- Tuples are immutable: `pt[0] = 999` (error)
- Why it matters: accidental changes can cause bugs; immutable structures are safer for fixed data.


````
nums = [10, 20, 30]
nums[0] = 999
print('mutable list:', nums)
pt = (4, 9)
print('tuple:', pt)
# Uncomment to see tuple immutability error:
# pt[0] = 999
````

## Common sequence operations
- `len(seq)`
- `x in seq`
- `seq + seq`
- `seq * n`


````
a = [1, 2, 3]
print('len(a)=', len(a))
print('2 in a:', 2 in a)
print('a + a =', a + a)
print('a * 3 =', a * 3)
````

## Common mistakes
- Off-by-one errors in slicing
- Confusing index (position) with value
- Trying to modify a tuple
- Forgetting the comma in a one-item tuple `(x,)`

## Iterating through a list
- Most common pattern: `for item in my_list:`
- When you need indices: `for i in range(len(my_list)):`
- Rule of thumb: prefer `for item in ...` unless you truly need the index.


> Image placeholder: iterating over a list vs using range(len(...)).


````
names = ['Ada', 'Linus', 'Grace']
for name in names:
    print('Hello,', name)

for i in range(len(names)):
    print(i, names[i])
````

## Core list methods (daily drivers)
- Add: `append`, `extend`, `insert`
- Remove: `pop`, `remove`, `del`


````
nums = []
nums.append(10)
nums.append(20)
nums.extend([30, 40])
nums.insert(1, 15)
print('after adds:', nums)
nums.remove(20)
p = nums.pop()  # pops last
print('popped:', p)
del nums[0]
print('after removes:', nums)
````

## Search and summary methods
- `index(x)` finds position (error if not found)
- `count(x)` counts occurrences
- Sorting:
  - `my_list.sort()` modifies list
  - `sorted(my_list)` returns a new list


````
vals = [3, 1, 3, 2]
print('index of 2:', vals.index(2))
print('count of 3:', vals.count(3))
print('sorted(vals):', sorted(vals))
vals.sort()
print('vals after sort():', vals)
````

## Nested lists (2D data)
- A list can contain other lists.
- Access pattern: `grid[row][col]`
- Useful for: matrices, simple spreadsheets, game boards.


> Image placeholder: 2D list and grid[row][col] access.


````
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]
print('grid[1][2] =', grid[1][2])
````

## Copying lists (important bug warning)
- Assignment creates an alias: `b = a`
- Copy with: `b = a.copy()` or `b = a[:]`
- Why it matters: changing one might change the other.


> Image placeholder: screenshot showing aliasing vs copying behavior.


````
a = [1, 2, 3]
b = a
c = a.copy()
a.append(99)
print('a:', a)
print('b (alias):', b)
print('c (copy):', c)
````

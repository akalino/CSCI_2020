## Create a module with `%%writefile`
Run the next cell to create `src/utils.py`.


```
%%writefile src/utils.py
"""Utility functions for Module 6 demos."""

def normalize_category(cat: str) -> str:
    """Strip whitespace and lowercase a category string."""
    return cat.strip().lower()

def safe_int(s: str, default: int = 0) -> int:
    """Convert string to int; return default if conversion fails."""
    try:
        return int(s)
    except (TypeError, ValueError):
        return default

def safe_float_money(s: str, default: float = 0.0) -> float:
    """Convert a money-like string to float (removes $ and commas)."""
    if s is None:
        return default
    try:
        cleaned = str(s).strip().replace('$', '').replace(',', '')
        if cleaned == '':
            return default
        return float(cleaned)
    except ValueError:
        return default
```

## Import styles
- `import utils` then call `utils.safe_int(...)`
- `from utils import safe_int` then call `safe_int(...)`

We’ll import from `src/` by adding it to `sys.path` for this demo.


```
# Import the module we just wrote
import sys
from pathlib import Path

SRC = str((Path.cwd() / 'src').resolve())
if SRC not in sys.path:
    sys.path.append(SRC)

import utils

print(utils.normalize_category('  Office '))
print(utils.safe_int('ten', default=0))
print(utils.safe_float_money(' $1,250.00 '))
```

## Create a script that imports your module
Run the next cell to create `src/main_demo.py`.


```
%%writefile src/main_demo.py
"""Main demo script that imports our utils module."""

from utils import normalize_category, safe_int, safe_float_money

def main():
    cat = normalize_category('  TECH ')
    qty = safe_int('3')
    price = safe_float_money('$9.99')
    total = qty * price
    print('category:', cat)
    print('qty:', qty)
    print('price:', price)
    print('total:', total)

if __name__ == '__main__':
    main()
```

## Run the script (Binder-friendly)
This is how you can run scripts without requiring students to use a terminal.


```
!python src/main_demo.py
```

## The `__main__` guard
- Code under `if __name__ == '__main__':` runs only when executed directly.
- It does NOT run when imported.
This keeps modules reusable.


## Common beginner issues
- `ModuleNotFoundError`: wrong folder or module not on path
- Naming collisions (don’t name a file `csv.py`, `json.py`, etc.)
- Confusing CWD (print it!)

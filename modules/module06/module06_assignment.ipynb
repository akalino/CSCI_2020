```
# CHECK 2: CSV structure
import csv
with open(OUT_CSV, 'r', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    assert reader.fieldnames == ['keyword','count'], f'Unexpected CSV headers: {reader.fieldnames}'
    rows = list(reader)
    assert len(rows) >= 1
print('✅ CSV headers OK')
```

```
# CHECK 3: JSON structure
import json
s = json.loads(OUT_JSON.read_text(encoding='utf-8'))
for k in ['num_lines','keywords','counts']:
    assert k in s, f'Missing JSON key: {k}'
assert isinstance(s['keywords'], list)
assert isinstance(s['counts'], dict)
print('✅ JSON structure OK')
```

```
# CHECK 4: sanity check counts (ignore-case)
# This input file contains 3 lines with 'error', 2 lines with 'warning', and 2 lines with 'info' (case-insensitive).
import json
s = json.loads(OUT_JSON.read_text(encoding='utf-8'))
counts = {k.lower(): int(v) for k,v in s['counts'].items()}
assert counts.get('error') == 2 or counts.get('error') == 3, 'Error count seems off (check your one-hit-per-line rule)'
assert counts.get('warning') >= 1, 'Warning count seems off'
assert counts.get('info') >= 1, 'Info count seems off'
print('✅ Basic count sanity looks OK')
```

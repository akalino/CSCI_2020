# Video 1 — Algorithmic Thinking: Instructions vs Algorithms

### Week 1 outcomes
- Define instructions, algorithms, and computation
- Translate problems into a set of explicit steps
- Formulate solutions in terms of algorithmic thinking


## How computers operate
- Computers are great at automating repetitive tasks.
- They need **very explicit instructions**.
- An **algorithm** formalizes instructions so a computer can follow them.


## Example: PB&J instructions
The point of this example is to show how natural language can be ambiguous.

**Try 1 (too vague):**
1. Buy bread, peanut butter, jelly
2. Take two slices of bread
3. Apply peanut butter to one slice
4. Apply jelly to the other
5. Put them together

**Question:** Is this good enough for a robot?


## Ambiguity vs specificity
- Humans infer meaning from experience.
- Computers do not.
- So we must be **specific** about exact actions.


## PB&J: Take 2 (more explicit)
This is what “more explicit” starts to look like:
- If the loaf is pre-sliced… else slice with a knife…
- Use a measuring spoon to remove **two tablespoons**…
- Spread into a **single, even layer**…

**Key idea:** We’re starting to see how detailed instructions need to be.


## Definitions
**Instructions:** a statement (natural language) of steps to accomplish a task.

**Algorithm:** a formally defined process/set of rules for calculations or problem-solving, typically by a computer.


## Implementing algorithms (semester goal)
- Learn to think algorithmically
- Turn algorithms into programs
- Use Python to solve problems / automate boring tasks


## Quick check
1) Rewrite this instruction to be unambiguous: *“Stir until done.”*

2) Which is closer to an algorithm: a paragraph description or numbered formal steps?

# Video 2 — Computation and I/O: The 3‑Box Model (Inputs → Process → Outputs)
Computers can execute fast, exact computations. We break problems into **Inputs**, **Process**, **Outputs**.


## The 3‑Box Model
**Inputs** → **Process** → **Outputs**

> Image placeholder: Insert the 3‑Box Model diagram from the slides here.


## Worked example: total meal cost
We’ll use the same structure for lots of problems.

**Inputs:**
- base cost of meal `b`
- tax rate `r`
- tip percentage `t`

**Process (explicit steps):**
1. Apply tax to base cost
2. Apply tip

**Output:** total cost


## Algorithmic thinking takeaway
- Identify inputs early (these become variables later)
- Break computation into distinct steps
- Track “state” of the computation to check for errors


## Other considerations
- Input/output types matter (percent vs decimal)
- Rounding rules matter (e.g., 2 decimals for dollars)
- We’ll formalize this with Python data types later


## Common algorithmic failures
1. Missing input → code cannot run
2. Wrong input/output type → unexpected result
3. Too much in one step → hard to debug; break it down

# Video 3 — Pseudocode
Pseudocode helps you design logic **before** worrying about Python syntax.


## Problem-solving process
**Algorithm → Pseudocode → Program**

> Image placeholder: Insert the process diagram from the slides here.


## Pseudocode style (course conventions)
- Use ALL CAPS for operations
- Keywords: `INPUT`, `OUTPUT`, `SET`, `IF/ELSE`, `FOR`
- Indentation matters (prepares you for Python)
- One action per line


## Pattern 1: Sequential ordering
```
INPUT 1,2
SET x = 1
SET y = 2
SET z = x + y
OUTPUT z
```
**Expected:** `z = 3`


## Pattern 2: IF-ELSE logic
```
INPUT 1,2
SET x = 1
SET y = 2
IF x > y:
    SET z = x
ELSE:
    SET z = y
OUTPUT z
```
**Expected:** `z = 2`


## Conditions for IF-ELSE
- The statement after `IF` is the **condition**
- It evaluates to True/False

**Try:** IF the input is negative, output an error, ELSE compute the square root.


## Pattern 3: Repetition (loops)
- Sometimes we repeat a step many times.
- In pseudocode, we can express a loop with `FOR`.


## Snack loop example
```
INPUT a bag of chips
SET n = number of chips
FOR each chip in the bag
    SET n = n - 1
OUTPUT 'bag is empty'
```
When `n = 0`, the loop stops.


## Wrap-up
- Pseudocode translates problem-solving descriptions to actions
- We use IPO language to frame actions
- We saw 3 patterns: Sequential, IF/ELSE, FOR loops

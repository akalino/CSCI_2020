# Video 5 — Python Execution Workflow + Errors
Programs don’t “run” — they execute line-by-line, moving state-to-state.


## Outcomes
- Describe (big picture) what happens when you run Python code
- Recognize what tracebacks tell us
- Identify and categorize common errors


## When you press “Run”
- You start with source code (file or notebook)
- The Python interpreter executes the code top-to-bottom
- Variables and functions are created as encountered
- Program state changes as lines execute


## Three classes of errors
1. **Syntax errors:** Python can’t parse the code
2. **Runtime exceptions:** code hits a problem while running
3. **Logic errors:** code runs but output is wrong


## Understanding tracebacks
- Python supplies a traceback
- Traceback tells file + line number + error type
- Exception message is a starting point for diagnosis


## Common errors (names you will see again)
- TypeError
- ValueError
- IndexError
- KeyError
These will make more sense once we learn data types in Module 2.


## Errors workflow
1. Find the line where the error occurred
2. Inspect the values being operated on
3. Change one thing at a time
4. Re-run code; repeat


## Debugging checklist
- What did you expect the output to be?
- What happened instead?
- Where do those diverge?
- Can you make a small reproducible example?
- Use print statements to inspect state

# Video 4 — Strings: Outputs, Comments, Basic Manipulation, and input()


## Print
- In the terminal: prints to standard out
- In notebooks: prints below the cell
- `print()` is built-in (standard library)
- Useful for debugging (inspect variable contents)


```
print('Hello from print')
x = 123
print('x is', x)
```

## General strings
- Strings use single or double quotes
- Double quotes avoid problems with contractions: "It's cold outside"
- Strings are made of characters


```
s1 = 'hello'
s2 = "It's cold outside"
print(s1)
print(s2)
print('First char of s1:', s1[0])
```

## Special characters
- Newline: `\n`
- Tab: `\t`
- Backslash: `\\`


```
print('line1\nline2')
print('col1\tcol2')
print('A backslash looks like this: \\')
```

## Comments
- Comments help document code
- Start with `#`
- Useful for disabling a line temporarily while testing


```
# This is a comment
x = 5  # also a comment
print(x)
```

## String methods
Strings have built-in utilities called **methods**.

Example: `"My string".upper()`


```
s = '  My string  '
print(s.upper())
print(s.lower())
print(s.strip())
```

## Inputs/Outputs
- `print()` outputs
- `input()` collects input (from terminal or notebook prompt)
- `input()` always returns a **string**


```
# Uncomment to try interactively
# name = input('What is your name? ')
# print('Hello,', name)
print('Note: input() returns a string')
```

## Re-usability
In Module 2 so far, we’ve written lines of code in sequence.
But we often have to re-write the same logic.

**There must be a better way...**


## DRY
**D**on’t **R**epeat **Y**ourself

Repeating code wastes effort and can introduce new errors.


## Functions
- A way to make code modular
- Small pieces you can use over and over
- Avoid copy/paste


## Re-visiting IPO for functions
- Inputs → function parameters
- Process → computations inside the function
- Output → return value


## Function structure (concept)
Restaurant bill example parameters:
- food cost, tax rate, tip percent

## Function tips
- Name functions clearly

- Document what it does

## Anatomy of a function (placeholder)
> Image placeholder: Insert the function anatomy snippet from the slides here.


```
# Example function (matches slide concept)
def compute_total_bill(food, tax_rate, tip_rate):
    """Compute total restaurant bill with tax and tip."""
    taxed_food = food + (food * tax_rate)
    tax_and_tip = taxed_food + (taxed_food * tip_rate)
    return tax_and_tip

print(compute_total_bill(20.0, 0.06, 0.20))
```

## Function formatting (PEP 8 basics)
- Indentation: 4 spaces (preferred)
- Keep lines to ~79 characters
- Two blank lines between top-level functions
- Whitespace: space after commas, around operators
We’ll enforce only a few standards for now.

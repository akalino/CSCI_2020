## Python programming language
- Python is a **high-level** language (abstracted away from hardware)
- Python is **object-oriented** (supports reusable, modular code)
- Python emphasizes readability: you can return to code later and understand it


## The Zen of Python
- Beauty and clarity of code
- One obvious way to do something
- Readability is important
- Simple rather than complex

We call code that follows these principles **pythonic**.


## Your first program
```python
print('Hello World!')
```
Things to note:
- `print()` is a built-in function
- `'Hello World!'` is a **string**


````
# Run this
print('Hello World!')
````

## Variables
**Definition:** A variable is a container that holds a particular value (number, string, or more complex object).


## Using variables
It’s pythonic to use lower-case variable names.

Example:
```python
x = 'Hello World!'
print(x)
```
`=` is called **assignment**.


````
x = 'Hello World!'
print(x)
````

## Reassignment
- Variables can be updated line-by-line.
- Updates can refer to the variable itself.

Example:
```python
x = 3
x = x + 1
```


````
x = 3
x = x + 1
print(x)
````

## Statements and expressions
**Statement:** standalone unit of execution that performs an action but does not evaluate to a value.

**Expression:** combination of values/variables/operators/function calls that evaluates to a single value.


````
# Expression examples
2 + 3 * 4
# Statement examples
x = 10
print(x)
````

# Video 2 — Expressions, Operators, and Numeric Types


## Combining variables
```python
x = 8
y = 2
x + y      # expression
z = x + y  # expression then assignment
```


````
x = 8
y = 2
print(x + y)
z = x + y
print('z =', z)
````

## Order of operations (PEMDAS)
- Parentheses
- Exponents
- Division
- Addition
- Subtraction
Evaluate left-to-right within the same level.


## Basic operators
- addition `+`
- subtraction `-`
- multiplication `*`
- division `/`
- exponents `**`


## Practice set
Define `w=16, x=3, y=4, z=3` then compute:
- `y - w`
- `(w/y) + (x/z)`
- `x**z - w`
- `z**(w/y)`
- `(w - y**2)`
- `((w - x**2) - 1) / x`


````
w, x, y, z = 16, 3, 4, 3
print('y-w =', y-w)
print('(w/y)+(x/z) =', (w/y) + (x/z))
print('x**z - w =', x**z - w)
print('z**(w/y) =', z**(w/y))
print('(w - y**2) =', (w - y**2))
print('((w - x**2) - 1) / x =', ((w - x**2) - 1) / x)
````

## What happens in division?
Even `4 / 2` gives `2.0`.

Two important numeric data types:
- `int`: whole numbers
- `float`: numbers with decimals


````
print(4/2)
print(type(4/2))
````

## Numeric type conversions
- Python often converts division results to float.
- You can cast:
  - `float(4) -> 4.0`
  - `int(4.2) -> 4`


````
print(float(4))
print(int(4.2))
````

# Video 3 — Advanced Operators + Assignment Shortcuts


## Advanced operators
- floor division `//`
- modulo (remainder) `%`
- comparisons: `<`, `>`, `<=`, `>=`, `==`


````
print('7 // 2 =', 7 // 2)
print('7 % 2 =', 7 % 2)
print('3 < 5:', 3 < 5)
print('3 == 3:', 3 == 3)
````

## Alternate assignments
- add-and-assign `+=`
- subtract-and-assign `-=`
- multiply-and-assign `*=`
- (and many others)


````
a = 10
a += 5
print('a after += 5:', a)
a *= 2
print('a after *= 2:', a)
a -= 3
print('a after -= 3:', a)
````

## Combining variable errors
It’s easy to mix `int` and `float`.
But mixing numeric types with strings can cause errors.


````
# Uncomment one at a time to see the error
# '3' + 4
# '3' * 2
print('Tip: strings behave differently with + and *')
````

## Absolute vs relative paths
- Absolute path: starts at the root of the drive
- Relative path: starts from your current working directory (CWD)
- Relative paths are convenient but depend on where you run code


````
# CWD basics
import os

print('os.getcwd():', os.getcwd())
print('Current folder files:', os.listdir(os.getcwd())[:10])
````

## Building paths safely
- Avoid hardcoding `/` or `\` in file paths
- Use `os.path.join(...)` or `pathlib.Path` to build paths
- Check existence with `os.path.exists` (or `Path.exists()`)


````
# Path building demo (os.path + pathlib)
import os
from pathlib import Path

p1 = os.path.join('data', 'sample.txt')
p2 = Path('data') / 'sample.txt'
print('os.path.join:', p1)
print('pathlib:', p2)
print('Exists right now?', Path(p1).exists())
````

## TXT files: read, process, write
- Always prefer:
```python
with open(path, 'r') as f:
    ...
```
- Reading patterns:
  - `f.read()` whole file
  - `f.readlines()` list of lines
  - `for line in f:` stream line-by-line (best for large files)
- Writing modes:
  - `'w'` overwrite
  - `'a'` append
> Image placeholders: insert read patterns + write modes slides.


````
# Create a sample TXT file in data/ (so demos always work)
from pathlib import Path

sample_txt = DATA_DIR / 'raw_notes.txt'
if not sample_txt.exists():
    sample_txt.write_text('  Apple\n\nBANANA  \n  kiwi\n', encoding='utf-8')
print('Sample file:', sample_txt)
print(sample_txt.read_text(encoding='utf-8'))
````

````
# Reading patterns demo
path = DATA_DIR / 'raw_notes.txt'

with open(path, 'r', encoding='utf-8') as f:
    whole = f.read()
print('--- f.read() ---')
print(whole)

with open(path, 'r', encoding='utf-8') as f:
    lines = f.readlines()
print('--- f.readlines() ---')
print(lines)

print('--- for line in f ---')
with open(path, 'r', encoding='utf-8') as f:
    for line in f:
        print(repr(line))
````

## Mini pipeline: clean and save
Goal: read an input TXT file, clean each line, and write cleaned output.
- Strip whitespace
- Remove blank lines
- Normalize case (lowercase)
> Image placeholder: pipeline diagram slide.


````
# Read → process → write pipeline
inp = DATA_DIR / 'raw_notes.txt'
out = OUTPUT_DIR / 'cleaned_notes.txt'

cleaned = []
with open(inp, 'r', encoding='utf-8') as f:
    for line in f:
        s = line.strip()
        if s == '':
            continue
        cleaned.append(s.lower())

with open(out, 'w', encoding='utf-8') as f:
    for item in cleaned:
        f.write(item + '\n')

print('Wrote:', out)
print(out.read_text(encoding='utf-8'))
````

## Common bugs + fixes
- `FileNotFoundError`: check CWD and print the path you're opening
- Output on one line: remember to write `\n`
- Accidentally overwrote a file: be careful with `'w'` mode

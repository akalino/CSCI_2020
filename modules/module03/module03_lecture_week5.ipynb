## If-else (recall from flowcharts)
- We check a condition.
- If the condition is true, we do one process.
- Otherwise (else), we do another.


## Anatomy of if-else syntax (placeholder)
> Image placeholder: Insert the slide code snippet showing `check_negative(num)`.


```
# Basic if/else example
def check_negative(num):
    if num < 0:
        return f"{num} is negative"
    else:
        return f"{num} is positive"

print(check_negative(-2))
print(check_negative(5))
```

## Boolean logic
- Boolean logic is logic involving True/False statements.
- We can combine conditions with `and` / `or`.
- We can negate with `not`.


## Truth tables (placeholder)
> Image placeholder: Insert the AND/OR truth table slide.


```
# Boolean logic demo
sunny = True
raining = False
print('sunny and raining:', sunny and raining)
print('sunny or raining:', sunny or raining)
print('not raining:', not raining)
```

## Another if-else example: type checking (placeholder)
> Image placeholder: Insert the slide code snippet showing `type_checker(num)`.


```
def type_checker(num):
    if (type(num) == int) or (type(num) == float):
        return 'Input was an int or float'
    else:
        return 'Input was not an int or float, retry'

print(type_checker(3))
print(type_checker(3.14))
print(type_checker('3'))
```

## Nested if-else
- Nesting = putting an `if/else` inside another `if/else`.
- Evaluations occur at each level.
- Start with the first `if` encountered, then move inward.


## Nested if-else example (placeholder)
> Image placeholder: Insert the nested if-else code slide.


```
def nested_if_else(x, y, z, w):
    if x == y:
        if z <= x + y:
            return w
        else:
            return y
    else:
        return z

print(nested_if_else(2, 2, 3, 99))
print(nested_if_else(2, 2, 10, 99))
print(nested_if_else(2, 5, 10, 99))
```

## elif (multi-branch decisions)
- Use `elif` when you need more than 2 branches.
- Structure: `if` → `elif` → `else`.


## Boolean + elif example (placeholder)
> Image placeholder: Insert the `boolean_if_elif(y, z, x='Stop')` slide.


```
def boolean_if_elif(y, z, x='Stop'):
    if x != 'Stop':
        return y
    elif x == 'Stop':
        return z
    else:
        return x + y

print(boolean_if_elif(1, 2, x='Go'))
print(boolean_if_elif(1, 2, x='Stop'))
```

### (Testing helper — do not edit)
The following helper lets us test `input()`-based functions by feeding fake inputs.


```
# CHECK HELPER (do not edit)
import builtins
from contextlib import contextmanager

@contextmanager
def fake_inputs(inputs):
    """Temporarily replace builtins.input to return values from inputs list."""
    it = iter(inputs)
    old_input = builtins.input
    builtins.input = lambda prompt='': next(it)
    try:
        yield
    finally:
        builtins.input = old_input

def check_equal(actual, expected, msg=''):
    if actual != expected:
        raise AssertionError(f"{msg}\nExpected: {expected}\nActual:   {actual}")
```

```
# CHECK Part 1 (do not edit)
with fake_inputs(['ten', '  42  ']):
    val = get_int('Enter an int: ')
check_equal(val, 42, 'Part 1 failed: get_int should accept spaced integer after rejecting bad input')
print('✅ Part 1 basic check passed')
```

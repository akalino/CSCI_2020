## Part A: Try/Except
- Programs can fail at runtime.
- We want to control that failure with a fallback option.
- Think of try/except like if/else: *try the code; if it fails, handle it*.


## Vocabulary
- **Exception:** error raised during execution (runtime)
- **Traceback:** the path of where the error occurred
- **Handling:** deciding what to do when an exception occurs
- **Fail fast:** sometimes you should not catch an error (let it surface)


## Common beginner exceptions
- ValueError
- TypeError
- NameError
- IndexError
- KeyError


## Anatomy of try/except (placeholder)
> Image placeholder: Insert slide with `compute_square(x)` capturing a TypeError.


```
def compute_square(x):
    try:
        y = x**2
        return y
    except TypeError:
        return 'Issue: captured a type error'

print(compute_square(5))
print(compute_square('five'))
```

## Keep try blocks small
- Put only the risky line(s) in the `try` block.
- Catch specific exceptions when possible.


```
s = '10'
try:
    x = int(s)  # risky line
except ValueError:
    print('Not an int')
else:
    print('x*2 =', x*2)
```

## Multiple except blocks
- Different errors → different responses
- Order matters: specific exceptions first


```
def safe_divide():
    try:
        a = int(input('Enter numerator: '))
        b = int(input('Enter denominator: '))
        print('a/b =', a/b)
    except ValueError:
        print('You must type whole numbers.')
    except ZeroDivisionError:
        print('Denominator cannot be 0.')

# Uncomment to try interactively
# safe_divide()
print('safe_divide() demo ready (uncomment to run)')
```

## Optional: else and finally
- `else` runs when no exception occurs
- `finally` runs no matter what (cleanup)
(You won’t need these often yet, but they exist.)


## Try/Except vs If-checks
- Use if-checks when you can validate before risky operations.
- Use try/except when failure is likely and checking is awkward.


## Debugging mindset with exceptions
- Read exception type and message
- Locate line number
- Identify values involved
- Fix one thing at a time


## Common mistakes
- Catching everything and printing "error" (not helpful)
- Wrapping whole program in one big try
- Continuing after an error without fixing bad state
- Not telling user what valid input looks like

## Part B: Loops, Iterators, and Generators
- Loops repeat tasks (avoid copy/paste).
- Support repeated computations: accumulate totals, process items, validate input.


## Two main loop types
- `while`: repeat while a condition is true (unknown number of reps)
- `for`: repeat over a sequence/iterable (do something for each item)


## Anatomy of a while loop
- condition
- body
- update
- if condition never becomes false → infinite loop


```
# While loop counter example
i = 0
while i < 5:
    print('i =', i)
    i += 1
```

## Common while-loop bugs
- Condition never changes
- Off-by-one (start/stop wrong)
- Updating the wrong variable


## Anatomy of a for loop
- `for item in sequence:`
- loop variable takes each value in order
- works with strings, range, lists, etc.


```
for ch in 'DATA':
    print(ch)
```

## range() tool
- `range(n)` gives 0..n-1
- `range(start, stop)` gives start..stop-1
- `range(start, stop, step)` increments by step


```
print(list(range(5)))
print(list(range(2, 7)))
print(list(range(10, 0, -2)))
```

## break and continue
- `break` exits loop early
- `continue` skips to next iteration


```
for n in range(1, 11):
    if n == 5:
        continue
    if n == 9:
        break
    print(n)
```

## Accumulation pattern
- initialize an accumulator (`total = 0`)
- loop and update (`total += x`)


```
total = 0
for i in range(1, 6):
    total += i
print('sum 1..5 =', total)
```

## Iterators and generators (conceptual)
- Iterators produce values one at a time (`iter`, `next`).
- Generators use `yield` to produce values lazily.
We only need the idea for now.


```
# Iterator demo
nums = [10, 20, 30]
it = iter(nums)
print(next(it))
print(next(it))
print(next(it))
```

```
# Generator demo
def countdown(n):
    while n > 0:
        yield n
        n -= 1

for x in countdown(5):
    print(x)
```
